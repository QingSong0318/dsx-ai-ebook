```python
import os
import subprocess

def convert_ipynb_to_md(folder_path):
    """
    遍历指定文件夹下所有的ipynb文件，并转换为md文件
    
    Args:
        folder_path: 要遍历的文件夹路径
    """
    # 统计转换数量
    converted_count = 0
    skipped_count = 0
    
    # 遍历文件夹
    for root, dirs, files in os.walk(folder_path):
        for filename in files:
            # 检查是否为ipynb文件
            if filename.endswith('.ipynb'):
                # 构建完整路径
                ipynb_path = os.path.join(root, filename)
                # 构建输出md文件路径（替换扩展名）
                md_path = os.path.join(root, filename[:-6] + '.md')
                
                # 检查md文件是否已存在
                if os.path.exists(md_path):
                    print(f"跳过已存在: {md_path}")
                    skipped_count += 1
                    continue
                
                try:
                    # 执行转换命令
                    print(f"转换中: {ipynb_path} -> {md_path}")
                    subprocess.run(
                        ['jupyter', 'nbconvert', '--to', 'markdown', ipynb_path],
                        check=True,
                        capture_output=True,
                        text=True
                    )
                    converted_count += 1
                except subprocess.CalledProcessError as e:
                    print(f"转换失败: {ipynb_path}")
                    print(f"错误信息: {e.stderr}")
    
    # 输出统计信息
    print(f"\n转换完成！")
    print(f"成功转换: {converted_count} 个文件")
    print(f"跳过已存在: {skipped_count} 个文件")
```

```python
convert_ipynb_to_md("./notebooks")
```
