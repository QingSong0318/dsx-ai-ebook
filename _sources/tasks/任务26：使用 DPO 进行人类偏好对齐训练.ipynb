# 【实战】使用 DPO 进行人类偏好对齐训练 

## 任务背景

在大模型训练过程中，监督微调（Supervised Fine-Tuning，SFT）通常用于让基础模型快速具备特定任务能力和基本指令对齐能力，是模型从"能生成"走向"会回答"的关键步骤。然而，SFT 让模型学会了"说人话"和遵循指令格式，但它本质上是在模仿训练数据中的文本分布，当我们想要在更泛化的场景下评判回答优劣时就会力不从心——它只能判断"像不像范例"，却无法区分两个合格回答中哪个更符合人类真实偏好。

因此需要引入直接偏好优化（Direct Preference Optimization，DPO），通过人类标注的"好/坏"对比数据，直接教会模型"答得好"而非仅仅"答得对"，从而在开放场景中输出真正受用户欢迎的结果。

## 任务目标

本次任务将在 Qwen2.5-0.5B 基座模型上，通过 DPO 训练让模型在问答中更倾向于生成人类偏好（带有 emoji）的结果，使得模型的问答结果更显生动、活泼、友好。

## 学习收获

完成本次任务，你将能够：

- **理解 DPO 的核心原理**：掌握如何利用人类偏好数据引导模型对齐优化，清晰辨析 DPO 与 SFT 在训练目标上的本质差异。
- **明确 DPO 训练的数据需求**：熟悉 DPO 所需的数据格式（prompt、chosen、rejected），并掌握构造高质量人类偏好对比数据的方法。
- **掌握 DPO 的标准训练流程**：理解 DPO 训练前为何需要先进行 SFT 微调——让模型具备基本的指令跟随能力，为后续偏好学习提供稳定可靠的起点。
- **了解 DPO 训练中的指标含义**：通过了解 `rewards/chosen`、`rewards/rejected` 等指标的含义，理解 DPO 训练的目标。
